```python

# ============================================================
# PROJECT: SPREAD LOCATOR
# A Statistical Distribution Analysis Model
# ============================================================

# 1. IMPORT REQUIRED LIBRARIES

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
import statsmodels.api as sm

from scipy.stats import (
    bernoulli, binom, poisson, lognorm,
    pareto, boxcox, normaltest, shapiro
)

import warnings
warnings.filterwarnings("ignore")

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)
plt.rcParams["font.size"] = 11

np.random.seed(42)

print("=" * 65)
print("SPREAD LOCATOR: STATISTICAL DISTRIBUTION ANALYSIS")
print("=" * 65)


# ============================================================
# 2. LOAD THE DATASET
# ============================================================

df = pd.read_excel("C:/Users/mansi/Downloads/spread_locator_dataset.xlsx")

print("\nDATASET LOADED SUCCESSFULLY")

print("\nFirst Five Records:")
display(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Information:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())

# Convert transaction date into datetime format
df["transaction_date"] = pd.to_datetime(
    df["transaction_date"]
)

# Remove duplicate records
df = df.drop_duplicates()

# Remove rows with missing required values
df = df.dropna(subset=[
    "transaction_amount",
    "transaction_date",
    "transaction_count",
    "transaction_status"
])

# Keep valid positive transaction amounts
df = df[df["transaction_amount"] > 0].copy()

amounts = df["transaction_amount"].astype(float)
weekly_counts = df["transaction_count"].astype(int)

print("\nCleaned Dataset Shape:")
print(df.shape)

print("\nDataset Summary:")
display(df.describe())


# ============================================================
# 3. EXPLORATORY DATA ANALYSIS
# ============================================================

print("\nTRANSACTION AMOUNT ANALYSIS")

print("Total Transactions:", len(df))
print("Average Transaction Amount: ₹", round(amounts.mean(), 2))
print("Median Transaction Amount: ₹", round(amounts.median(), 2))
print("Minimum Transaction Amount: ₹", round(amounts.min(), 2))
print("Maximum Transaction Amount: ₹", round(amounts.max(), 2))
print("Standard Deviation: ₹", round(amounts.std(), 2))
print("Skewness:", round(amounts.skew(), 3))
print("Kurtosis:", round(amounts.kurtosis(), 3))

# Transaction amount distribution
plt.figure(figsize=(10, 5))
sns.histplot(amounts, bins=25, kde=True)
plt.title("Distribution of Transaction Amounts")
plt.xlabel("Transaction Amount (₹)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

print("""
Interpretation:
The histogram shows the distribution of customer transaction
amounts. The mean, median and skewness help us understand
the spread and shape of the transaction data.
""")


# ============================================================
# 4. BERNOULLI DISTRIBUTION
# ============================================================

print("\n" + "=" * 65)
print("4. BERNOULLI DISTRIBUTION")
print("=" * 65)

# Success = 1, Fail = 0
df["transaction_occurrence"] = (
    df["transaction_status"].str.strip().str.lower() == "success"
).astype(int)

success_probability = df["transaction_occurrence"].mean()
failure_probability = 1 - success_probability

print("Probability of Success:", round(success_probability, 4))
print("Probability of Failure:", round(failure_probability, 4))

print("\nBernoulli Distribution:")
print("P(X = 1) =", round(success_probability, 4))
print("P(X = 0) =", round(failure_probability, 4))

x = np.array([0, 1])

bernoulli_probabilities = bernoulli.pmf(
    x, success_probability
)

plt.figure(figsize=(8, 5))
plt.bar(
    ["Failure (0)", "Success (1)"],
    bernoulli_probabilities,
    width=0.5
)
plt.title("Bernoulli Distribution of Transaction Status")
plt.xlabel("Transaction Outcome")
plt.ylabel("Probability")
plt.ylim(0, 1)
plt.tight_layout()
plt.show()

print("""
Interpretation:
Bernoulli distribution represents two possible outcomes:
Success and Failure.

The probability of success is calculated from the actual
transaction status column of the dataset.
""")


# ============================================================
# 5. BINOMIAL DISTRIBUTION
# ============================================================

print("\n" + "=" * 65)
print("5. BINOMIAL DISTRIBUTION")
print("=" * 65)

# Weekly transaction count
n = int(weekly_counts.max())

# Estimate probability of one transaction
p = weekly_counts.mean() / n

print("Maximum Observed Weekly Count (n):", n)
print("Estimated Probability (p):", round(p, 4))
print("Average Weekly Transaction Count:", round(weekly_counts.mean(), 2))

if 0 <= p <= 1 and n > 0:

    x_binomial = np.arange(0, n + 1)

    binomial_probabilities = binom.pmf(
        x_binomial, n, p
    )

    plt.figure(figsize=(9, 5))
    plt.bar(
        x_binomial,
        binomial_probabilities,
        color="steelblue",
        label="Binomial Probability"
    )

    plt.title("Binomial Distribution of Weekly Transaction Counts")
    plt.xlabel("Number of Transactions")
    plt.ylabel("Probability")
    plt.legend()
    plt.tight_layout()
    plt.show()

    print("\nExpected Weekly Count:", round(n * p, 2))
    print("Binomial Variance:", round(n * p * (1 - p), 2))

print("""
Interpretation:
Binomial distribution models the number of occurrences
in a fixed number of trials.

Here, weekly transaction counts are analysed using the
observed maximum count as the assumed number of trials.
This is a modelling assumption because the actual maximum
number of possible weekly transactions is not provided.
""")


# ============================================================
# 6. POISSON DISTRIBUTION
# ============================================================

print("\n" + "=" * 65)
print("6. POISSON DISTRIBUTION")
print("=" * 65)

# Count transactions per day
daily_transactions = df.groupby(
    df["transaction_date"].dt.normalize()
).size()

# Include dates with zero transactions
all_dates = pd.date_range(
    start=df["transaction_date"].min().normalize(),
    end=df["transaction_date"].max().normalize(),
    freq="D"
)

daily_transactions = daily_transactions.reindex(
    all_dates, fill_value=0
)

lambda_value = daily_transactions.mean()

print("Average Transactions Per Day (Lambda):",
      round(lambda_value, 4))

print("Total Days Analysed:", len(daily_transactions))

x_poisson = np.arange(
    0,
    int(daily_transactions.max()) + 1
)

poisson_probabilities = poisson.pmf(
    x_poisson, lambda_value
)

plt.figure(figsize=(10, 5))

plt.hist(
    daily_transactions,
    bins=np.arange(
        -0.5,
        daily_transactions.max() + 1.5,
        1
    ),
    density=True,
    alpha=0.6,
    label="Observed Daily Transactions"
)

plt.plot(
    x_poisson,
    poisson_probabilities,
    "ro-",
    label="Poisson Distribution"
)

plt.title("Poisson Distribution of Daily Transactions")
plt.xlabel("Number of Transactions Per Day")
plt.ylabel("Probability")
plt.legend()
plt.tight_layout()
plt.show()

print("""
Interpretation:
Poisson distribution models the number of events occurring
within a fixed time interval.

Here, it is used to analyse the number of transactions
occurring per day.

Lambda represents the average number of daily transactions.
""")


# ============================================================
# 7. LOG-NORMAL DISTRIBUTION
# ============================================================

print("\n" + "=" * 65)
print("7. LOG-NORMAL DISTRIBUTION")
print("=" * 65)

# Fit Log-Normal distribution
lognorm_shape, lognorm_loc, lognorm_scale = lognorm.fit(
    amounts, floc=0
)

lognorm_pdf = lognorm.pdf(
    np.sort(amounts),
    lognorm_shape,
    loc=lognorm_loc,
    scale=lognorm_scale
)

print("Log-Normal Distribution Parameters:")
print("Shape:", round(lognorm_shape, 4))
print("Location:", round(lognorm_loc, 4))
print("Scale:", round(lognorm_scale, 4))

plt.figure(figsize=(10, 5))

sns.histplot(
    amounts,
    bins=25,
    stat="density",
    color="skyblue",
    label="Actual Data"
)

x_values = np.linspace(amounts.min(), amounts.max(), 300)

plt.plot(
    x_values,
    lognorm.pdf(
        x_values,
        lognorm_shape,
        loc=lognorm_loc,
        scale=lognorm_scale
    ),
    "r-",
    linewidth=2,
    label="Log-Normal Fit"
)

plt.title("Log-Normal Distribution of Transaction Amounts")
plt.xlabel("Transaction Amount (₹)")
plt.ylabel("Density")
plt.legend()
plt.tight_layout()
plt.show()

print("""
Interpretation:
Log-Normal distribution is used for positive and right-skewed
data.

It helps analyse transaction amounts where a small number
of transactions may have very high values.
""")


# ============================================================
# 8. POWER LAW DISTRIBUTION
# ============================================================

print("\n" + "=" * 65)
print("8. POWER LAW DISTRIBUTION")
print("=" * 65)

# Pareto distribution as a power-law model
pareto_shape, pareto_loc, pareto_scale = pareto.fit(
    amounts, floc=0
)

print("Power Law (Pareto) Parameters:")
print("Shape:", round(pareto_shape, 4))
print("Location:", round(pareto_loc, 4))
print("Scale:", round(pareto_scale, 4))

plt.figure(figsize=(10, 5))

sns.histplot(
    amounts,
    bins=25,
    stat="density",
    color="lightgreen",
    label="Actual Data"
)

plt.plot(
    x_values,
    pareto.pdf(
        x_values,
        pareto_shape,
        loc=pareto_loc,
        scale=pareto_scale
    ),
    "r-",
    linewidth=2,
    label="Power Law (Pareto) Fit"
)

plt.title("Power Law Distribution of Transaction Amounts")
plt.xlabel("Transaction Amount (₹)")
plt.ylabel("Density")
plt.legend()
plt.tight_layout()
plt.show()

print("""
Interpretation:
Power Law distribution describes data in which a few
observations may have very large values.

It is useful for examining high-value transactions
and extreme spending behaviour.
""")


# ============================================================
# 9. COMPARE LOG-NORMAL AND POWER LAW DISTRIBUTIONS
# ============================================================

print("\n" + "=" * 65)
print("9. DISTRIBUTION FIT COMPARISON")
print("=" * 65)

# Calculate log likelihood
lognorm_log_likelihood = np.sum(
    lognorm.logpdf(
        amounts,
        lognorm_shape,
        loc=lognorm_loc,
        scale=lognorm_scale
    )
)

pareto_log_likelihood = np.sum(
    pareto.logpdf(
        amounts,
        pareto_shape,
        loc=pareto_loc,
        scale=pareto_scale
    )
)

# AIC = 2k - 2 log likelihood
lognorm_aic = 2 * 2 - 2 * lognorm_log_likelihood
pareto_aic = 2 * 2 - 2 * pareto_log_likelihood

# Kolmogorov-Smirnov tests
lognorm_ks = stats.kstest(
    amounts,
    "lognorm",
    args=(lognorm_shape, lognorm_loc, lognorm_scale)
)

pareto_ks = stats.kstest(
    amounts,
    "pareto",
    args=(pareto_shape, pareto_loc, pareto_scale)
)

comparison = pd.DataFrame({
    "Distribution": ["Log-Normal", "Power Law (Pareto)"],
    "AIC": [lognorm_aic, pareto_aic],
    "KS Statistic": [lognorm_ks.statistic, pareto_ks.statistic],
    "KS P-Value": [lognorm_ks.pvalue, pareto_ks.pvalue]
})

print("\nDistribution Comparison:")
display(comparison.round(4))

if lognorm_aic < pareto_aic:
    best_distribution = "Log-Normal"
else:
    best_distribution = "Power Law (Pareto)"

print("Best Model by AIC:", best_distribution)

print("""
Interpretation:
A lower AIC indicates a better relative model fit.
A smaller KS statistic indicates a smaller difference
between the fitted distribution and observed data.

The KS p-values are approximate because distribution
parameters were estimated from the same dataset.
""")


# ============================================================
# 10. Q-Q PLOT AND NORMALITY TEST
# ============================================================

print("\n" + "=" * 65)
print("10. Q-Q PLOT AND NORMALITY TEST")
print("=" * 65)

# Q-Q plot for original transaction amounts
sm.qqplot(
    amounts,
    line="s"
)

plt.title("Q-Q Plot of Original Transaction Amounts")
plt.xlabel("Theoretical Quantiles")
plt.ylabel("Sample Quantiles")
plt.tight_layout()
plt.show()

# Normality tests
normal_stat, normal_p = normaltest(amounts)

shapiro_stat, shapiro_p = shapiro(amounts)

print("D'Agostino Normality Test:")
print("Statistic:", round(normal_stat, 4))
print("P-Value:", round(normal_p, 6))

print("\nShapiro-Wilk Test:")
print("Statistic:", round(shapiro_stat, 4))
print("P-Value:", round(shapiro_p, 6))

if normal_p < 0.05:
    print("\nConclusion: Transaction amounts are not normally distributed.")
else:
    print("\nConclusion: Normality cannot be rejected.")

print("""
Interpretation:
In a Q-Q plot, points close to the straight reference line
indicate approximate normality.

Large deviations from the line indicate that the data
may not follow a normal distribution.

A p-value below 0.05 indicates evidence against normality.
""")


# ============================================================
# 11. BOX-COX TRANSFORMATION
# ============================================================

print("\n" + "=" * 65)
print("11. BOX-COX TRANSFORMATION")
print("=" * 65)

# Box-Cox requires strictly positive values
transformed_amounts, boxcox_lambda = boxcox(amounts)

print("Box-Cox Lambda:", round(boxcox_lambda, 4))

print("Original Skewness:", round(amounts.skew(), 4))

print(
    "Transformed Skewness:",
    round(pd.Series(transformed_amounts).skew(), 4)
)

# Compare original and transformed data
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(
    amounts,
    bins=25,
    kde=True,
    ax=axes[0]
)

axes[0].set_title("Original Transaction Amounts")
axes[0].set_xlabel("Transaction Amount (₹)")

sns.histplot(
    transformed_amounts,
    bins=25,
    kde=True,
    ax=axes[1]
)

axes[1].set_title("Box-Cox Transformed Amounts")
axes[1].set_xlabel("Transformed Values")

plt.tight_layout()
plt.show()

# Q-Q plot after transformation
sm.qqplot(
    transformed_amounts,
    line="s"
)

plt.title("Q-Q Plot After Box-Cox Transformation")
plt.tight_layout()
plt.show()

# Normality test after transformation
transformed_normal_stat, transformed_normal_p = normaltest(
    transformed_amounts
)

print("\nNormality Test After Box-Cox Transformation:")
print("Statistic:", round(transformed_normal_stat, 4))
print("P-Value:", round(transformed_normal_p, 6))

print("""
Interpretation:
Box-Cox transformation is used to reduce skewness and
make data more suitable for statistical analysis.

The original and transformed distributions are compared
using histograms, skewness and Q-Q plots.
""")


# ============================================================
# 12. Z-SCORE AND PROBABILITY ABOVE ₹5000
# ============================================================

print("\n" + "=" * 65)
print("12. Z-SCORE AND PROBABILITY ANALYSIS")
print("=" * 65)

mean_amount = amounts.mean()
std_amount = amounts.std()

# Calculate Z-scores
df["Z_Score"] = (
    amounts - mean_amount
) / std_amount

print("Mean Transaction Amount: ₹", round(mean_amount, 2))
print("Standard Deviation: ₹", round(std_amount, 2))

print("\nFirst Five Z-Scores:")
display(df[[
    "transaction_amount",
    "Z_Score"
]].head())

# Calculate Z-score for ₹5000
threshold = 5000

z_score_5000 = (
    threshold - mean_amount
) / std_amount

# Probability using normal distribution
normal_probability = stats.norm.sf(z_score_5000)

# Probability using fitted log-normal distribution
lognormal_probability = lognorm.sf(
    threshold,
    lognorm_shape,
    loc=lognorm_loc,
    scale=lognorm_scale
)

# Actual probability from dataset
empirical_probability = np.mean(amounts > threshold)

print("\nZ-Score for ₹5000:", round(z_score_5000, 4))

print(
    "Normal Model Probability of Amount > ₹5000:",
    round(normal_probability, 4)
)

print(
    "Log-Normal Model Probability of Amount > ₹5000:",
    round(lognormal_probability, 4)
)

print(
    "Actual Dataset Probability of Amount > ₹5000:",
    round(empirical_probability, 4)
)

print(
    "Actual Number of Transactions Above ₹5000:",
    int((amounts > threshold).sum())
)

print("""
Interpretation:
Z-score indicates how many standard deviations a value
is away from the mean.

The probability of a transaction exceeding ₹5000 is
calculated using the normal model, fitted log-normal model
and actual dataset proportion.

Because transaction amounts may be skewed, the empirical
and log-normal probabilities are also important comparisons.
""")


# ============================================================
# 13. PROBABILITY DENSITY FUNCTION (PDF)
# ============================================================

print("\n" + "=" * 65)
print("13. PROBABILITY DENSITY FUNCTION")
print("=" * 65)

plt.figure(figsize=(10, 5))

sns.histplot(
    amounts,
    bins=25,
    stat="density",
    color="skyblue",
    label="Observed Data"
)

plt.plot(
    x_values,
    lognorm.pdf(
        x_values,
        lognorm_shape,
        loc=lognorm_loc,
        scale=lognorm_scale
    ),
    color="red",
    linewidth=2,
    label="Log-Normal PDF"
)

plt.title("Probability Density Function of Transaction Amounts")
plt.xlabel("Transaction Amount (₹)")
plt.ylabel("Probability Density")
plt.legend()
plt.tight_layout()
plt.show()

print("""
Interpretation:
PDF represents the probability density of continuous
transaction amounts.

The area under the PDF curve between two values
represents the probability of a transaction falling
within that interval.
""")


# ============================================================
# 14. CUMULATIVE DISTRIBUTION FUNCTION (CDF)
# ============================================================

print("\n" + "=" * 65)
print("14. CUMULATIVE DISTRIBUTION FUNCTION")
print("=" * 65)

plt.figure(figsize=(10, 5))

plt.plot(
    x_values,
    lognorm.cdf(
        x_values,
        lognorm_shape,
        loc=lognorm_loc,
        scale=lognorm_scale
    ),
    color="green",
    linewidth=2,
    label="Log-Normal CDF"
)

# Empirical CDF
sorted_amounts = np.sort(amounts)

empirical_cdf = (
    np.arange(1, len(sorted_amounts) + 1)
    / len(sorted_amounts)
)

plt.step(
    sorted_amounts,
    empirical_cdf,
    where="post",
    color="blue",
    label="Empirical CDF"
)

plt.title("Cumulative Distribution Function of Transaction Amounts")
plt.xlabel("Transaction Amount (₹)")
plt.ylabel("Cumulative Probability")
plt.legend()
plt.tight_layout()
plt.show()

print("""
Interpretation:
CDF represents the cumulative probability that a transaction
amount is less than or equal to a particular value.

It helps understand the percentage of transactions
falling below a given transaction amount.
""")


# ============================================================
# 15. FINAL STATISTICAL SUMMARY
# ============================================================

print("\n" + "=" * 65)
print("15. FINAL STATISTICAL SUMMARY")
print("=" * 65)

summary = pd.DataFrame({
    "Measure": [
        "Total Transactions",
        "Mean Transaction Amount",
        "Median Transaction Amount",
        "Standard Deviation",
        "Minimum Amount",
        "Maximum Amount",
        "Skewness",
        "Daily Average Transactions",
        "Success Probability",
        "Probability Above ₹5000",
        "Box-Cox Lambda"
    ],

    "Value": [
        len(df),
        round(amounts.mean(), 2),
        round(amounts.median(), 2),
        round(amounts.std(), 2),
        round(amounts.min(), 2),
        round(amounts.max(), 2),
        round(amounts.skew(), 4),
        round(lambda_value, 2),
        round(success_probability, 4),
        round(empirical_probability, 4),
        round(boxcox_lambda, 4)
    ]
})

display(summary)


# ============================================================
# 16. FINAL CONCLUSION
# ============================================================

print("\n" + "=" * 65)
print("FINAL PROJECT CONCLUSION")
print("=" * 65)

print(f"""
1. The dataset contains {len(df)} valid transaction records.

2. The average transaction amount is ₹{amounts.mean():.2f}.

3. The median transaction amount is ₹{amounts.median():.2f}.

4. The transaction success probability is
   {success_probability:.4f}.

5. The average number of daily transactions is
   {lambda_value:.2f}.

6. The best model by AIC among the two fitted amount
   distributions is {best_distribution}.

7. The probability of a transaction exceeding ₹5000
   in the actual dataset is {empirical_probability:.4f}.

8. Box-Cox transformation was applied to reduce skewness
   and improve the distribution of transaction amounts.

OVERALL CONCLUSION:

The Spread Locator project applies Bernoulli, Binomial,
Poisson, Log-Normal and Power Law distributions to analyse
customer transaction behaviour.

Q-Q plots, normality tests, Z-scores, Box-Cox transformation,
PDF and CDF provide additional statistical insights.

These results can help an e-commerce platform understand
transaction patterns, customer spending behaviour and
high-value transactions.

The fitted distribution comparison helps identify which
model describes transaction amounts better among the
models considered.
""")

print("=" * 65)
print("=" * 65)
```
